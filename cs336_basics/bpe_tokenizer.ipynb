```python
import regex as re

PAT = r"""'(?:[sdmt]|ll|ve|re)| ?\p{L}+| ?\p{N}+| ?[^\s\p{L}\p{N}]+|\s+(?!\S)|\s+"""

exp_text = """
low low low low low
lower lower widest widest widest
newest newest newest newest newest newest
"""
```

```python
from collections import Counter

#init word counts
init_word_counts = Counter(exp_text.split())
print(init_word_counts)

counts:dict[tuple[bytes, ...], int] = {}

for word, frequency in init_word_counts.items():
    encoded = word.encode("utf-8")
    pieces = tuple(encoded[i:i+1] for i in range(len(encoded)))
    counts[pieces] = frequency

print(counts)
```

```python
def update_counts(current_counts:Counter):
    #find current max num corpus
    pair_counts = Counter()
    for pieces, frequency in current_counts.items():
        for pair in zip(pieces, pieces[1:]):
            pair_counts[pair] += frequency

    print(pair_counts)
    
    if not pair_counts: return pair_counts.copy()
    
    best_pair = pair_counts.most_common(1)[0][0]
    new_counts = Counter()

    #merge corpus    
    for pieces, frequency in current_counts.items():
        merged:list[bytes] = []
        i = 0
        
        while i < len(pieces):
            if i + 1 < len(pieces) and (pieces[i], pieces[i+1]) == best_pair:
                merged.append(pieces[i] + pieces[i+1])
                i += 2
            else:
                merged.append(pieces[i])
                i += 1
        
        new_counts[tuple(merged)]+= frequency
    
    return new_counts
```

```python
for i in range(6):
    counts = update_counts(counts)
    print(f"{i}th merge result : {counts}")
```
